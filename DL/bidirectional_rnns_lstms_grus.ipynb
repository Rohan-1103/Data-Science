```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense, LSTM, GRU, Input, Bidirectional
```

```python
# Load the dataset
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=20000)

x_train = pad_sequences(x_train, maxlen=100)
x_test = pad_sequences(x_test, maxlen=100)
```

```python
# Define the Bidirectional RNN model
model_birnn = Sequential([
    Input(shape = (100, )),
    Embedding(20000, 32),
    Bidirectional(SimpleRNN(5, return_sequences=True)),
    Bidirectional(SimpleRNN(5)),
    Dense(1, activation='sigmoid')
])
```

```python
model_birnn.summary()
```

```python
model_birnn.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
history_birnn = model_birnn.fit(x_train, y_train, epochs = 10, batch_size=32, validation_split=0.2)
```

```python
# Define the Bidirectional LSTM model
model_bilstm = Sequential([
    Input(shape = (100, )),
    Embedding(20000, 32),
    Bidirectional(LSTM(5, return_sequences=True)),
    Bidirectional(LSTM(5)),
    Dense(1, activation='sigmoid')
])
```

```python
model_bilstm.summary()
```

```python
model_bilstm.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
history_bilstm = model_bilstm.fit(x_train, y_train, batch_size = 32, epochs = 10, validation_split = 0.20)
```

```python
# Define the bidirectional GRU model
model_bigru = Sequential([
    Input(shape = (100, )),
    Embedding(20000, 32),
    Bidirectional(GRU(5, return_sequences=True)),
    Bidirectional(GRU(5)),
    Dense(1, activation='sigmoid')
])
```

```python
model_bigru.summary()
```

```python
model_bigru.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
history_bigru = model_bigru.fit(x_train, y_train, batch_size=32, epochs = 10, validation_split=0.20)
```
